```python
from sklearn.impute import KNNImputer


cols_numeriques = ['Age', 'RestingBP', 'Cholesterol', 'FastingBS', 'MaxHR', 'Oldpeak']

cols_a_imputer = ['RestingBP', 'Cholesterol']

cible = 'HeartDisease'

df_numerique = df[cols_numeriques].copy()

df_complet = df_numerique.dropna(subset=cols_a_imputer).copy()

df_manquant = df_numerique[df_numerique[cols_a_imputer].isnull().any(axis=1)].copy()

print(f"Lignes complètes : {len(df_complet)}")
print(f"Lignes avec des NaN : {len(df_manquant)}")
print("")

k_a_tester = [3, 5, 7, 9, 10, 11]
resultats_par_k = {k: [] for k in k_a_tester}

for test in range(1, 4):
    print(f" Test {test} (random seed = {42+test})")
    
    taille_echantillon = int(0.2 * len(df_complet))
    echantillon = df_complet.sample(n=taille_echantillon, random_state=42+test)
    
    df_reference = df_complet.drop(echantillon.index)
    
    echantillon_masque = echantillon.copy()
    echantillon_masque[cols_a_imputer] = np.nan
    
    df_a_imputer = pd.concat([df_reference, echantillon_masque], axis=0)
    
    vraies_valeurs = echantillon[cols_a_imputer].copy()
    
    print(f" Échantillon test : {len(echantillon)} lignes")
    
    for k in k_a_tester:
        imputer = KNNImputer(n_neighbors=k, weights='uniform', metric='nan_euclidean')
        df_impute_temp = pd.DataFrame(
            imputer.fit_transform(df_a_imputer),
            columns=cols_numeriques
        )
        
        debut_echantillon = len(df_reference)
        echantillon_impute = df_impute_temp.iloc[debut_echantillon:].copy()
        
        bp_impute = echantillon_impute['RestingBP'].values
        chol_impute = echantillon_impute['Cholesterol'].values
        
        bp_reel = vraies_valeurs['RestingBP'].values
        chol_reel = vraies_valeurs['Cholesterol'].values
        
        mae_bp = np.mean(np.abs(bp_impute - bp_reel))
        mae_chol = np.mean(np.abs(chol_impute - chol_reel))
        mae_moyenne = (mae_bp + mae_chol) / 2
        
        resultats_par_k[k].append(mae_moyenne)
        print(f"    K={k} -> MAE moyenne = {mae_moyenne:.2f} (BP: {mae_bp:.2f} | Chol: {mae_chol:.2f})")

print("\n RÉSULTATS FINAUX (moyenne sur les 3 tests) ")
meilleur_k = None
meilleure_mae = 9999

for k in k_a_tester:
    mae_moyenne = np.mean(resultats_par_k[k])
    print(f"K={k} -> MAE moyenne = {mae_moyenne:.2f}")
    if mae_moyenne < meilleure_mae:
        meilleure_mae = mae_moyenne
        meilleur_k = k

print(f" MEILLEUR K : K = {meilleur_k}  (MAE moyenne = {meilleure_mae:.2f})")

df_total = pd.concat([df_complet, df_manquant], axis=0)

imputer_final = KNNImputer(n_neighbors=meilleur_k, weights='uniform', metric='nan_euclidean')
df_impute_final = pd.DataFrame(
    imputer_final.fit_transform(df_total),
    columns=cols_numeriques
)

nb_nan = df_impute_final[cols_a_imputer].isnull().sum().sum()
print(f"NaN restants après imputation : {nb_nan}")

df_original = df.copy()
df_original[cols_numeriques] = df_impute_final

df = df_original
```
